
# Bag of Words (BoW) in NLP

**Bag of Words (BoW)** is a text representation technique used in Natural Language Processing (NLP) to convert text into numerical vectors based on the frequency (count) of words in a document.

In simple words, Bag of Words converts sentences into numbers by counting how many times each word appears.

It is called a *Bag of Words* because it considers word frequency but ignores grammar and word order.

## 1. How Does Bag of Words Work?

Suppose we have three sentences:

- Sentence 1: I love Python
- Sentence 2: I love Machine Learning
- Sentence 3: Python is amazing

### Step 1: Create a Vocabulary

Collect all unique words from the sentences:

```text
I, love, Python, Machine, Learning, is, amazing
```

### Step 2: Count the Frequency of Each Word

| Sentence | I | love | Python | Machine | Learning | is | amazing |
|---|---|---|---|---|---|---|---|
| Sentence 1 | 1 | 1 | 1 | 0 | 0 | 0 | 0 |
| Sentence 2 | 1 | 1 | 0 | 1 | 1 | 0 | 0 |
| Sentence 3 | 0 | 0 | 1 | 0 | 0 | 1 | 1 |

Each row represents a sentence, and each column represents a unique word.

## 2.Example: Python Implementation Using CountVectorizer

We use `CountVectorizer` from Scikit-learn to implement Bag of Words.

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample sentences
sentences = [
    "I love Python",
    "I love Machine Learning",
    "Python is amazing"
]

# Create CountVectorizer
cv = CountVectorizer()

# Apply Bag of Words
vectors = cv.fit_transform(sentences)

# Display vocabulary
print("Vocabulary:", cv.vocabulary_)

# Display numerical vectors
print("\nEncoded Data:")
print(vectors.toarray())
```

### Output

```text
Vocabulary: {'love': 4, 'python': 5, 'machine': 3,
             'learning': 2, 'is': 1, 'amazing': 0}

Encoded Data:
[[0 0 0 0 1 1]
 [0 0 1 1 1 0]
 [1 1 0 0 0 1]]
```

Note: `CountVectorizer` converts words to lowercase and ignores common English stopwords such as `I` by default.

## 3. Understanding CountVectorizer

| Function | Purpose |
|---|---|
| `CountVectorizer()` | Creates the vectorizer |
| `fit()` | Learns the vocabulary |
| `transform()` | Converts text into numerical vectors |
| `fit_transform()` | Learns vocabulary and converts text |
| `toarray()` | Converts sparse matrix into a normal array |

## 4. Advantages of Bag of Words

1. Easy to understand and implement.
2. Converts text into numerical data.
3. Captures word frequency.
4. Useful for text classification and spam detection.

## 5. Disadvantages of Bag of Words

1. Ignores the order of words.
2. Does not understand the meaning of words.
3. Creates large vectors for large vocabularies.
4. Does not capture relationships between similar words.

For example:

- "I love Python"
- "Python love I"

Both sentences produce the same Bag of Words representation.

## 6. Applications of Bag of Words

- Spam Detection
- Sentiment Analysis
- Text Classification
- Document Similarity
- News Categorization

## 7. Conclusion

Bag of Words (BoW) is a text representation technique in NLP that converts text into numerical vectors based on the frequency of unique words in a document. It creates a vocabulary of unique words and counts how many times each word appears in each sentence or document. Although it is simple and useful for text classification, it ignores word order and semantic meaning.

## 8. One-Hot Encoding vs Bag of Words

| Feature | One-Hot Encoding | Bag of Words |
|---|---|---|
| Purpose | Represents individual categories or words | Represents entire documents |
| Values | 0 and 1 | Word counts |
| Frequency | Does not represent frequency | Captures frequency |
| Output | Binary vector per category | Vector per document |
| Word order | Ignored | Ignored |

**Important:** One-Hot Encoding represents individual words, whereas Bag of Words represents a sentence or document using the frequency of its vocabulary words.
