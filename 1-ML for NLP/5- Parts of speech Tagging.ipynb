
# POS Tagging (Parts of Speech Tagging) in NLP

**POS Tagging (Part-of-Speech Tagging)** is the process of identifying and assigning grammatical categories to each word in a sentence based on its role and context.

These grammatical categories include nouns, verbs, adjectives, adverbs, pronouns, prepositions, conjunctions, and determiners.

In simple words, POS Tagging tells us what role each word plays in a sentence.

## 1. Parts of Speech

| Part of Speech | Meaning | Examples |
|---|---|---|
| Noun (NN) | Name of a person, place, or thing | Ishwar, Delhi, Book |
| Verb (VB) | Action or state | Run, Eat, Play |
| Adjective (JJ) | Describes a noun | Beautiful, Good, Tall |
| Adverb (RB) | Describes a verb, adjective, or adverb | Quickly, Slowly |
| Pronoun (PRP) | Replaces a noun | He, She, They |
| Preposition (IN) | Shows relationship between words | In, On, At |
| Conjunction (CC) | Joins words or sentences | And, But, Or |
| Determiner (DT) | Introduces a noun | The, A, An |
| Interjection (UH) | Expresses emotion | Wow!, Oh! |

## 2. Example of POS Tagging

Consider the sentence:

**"The boy is playing cricket."**

After POS Tagging:

| Word | POS Tag | Meaning |
|---|---|---|
| The | DT | Determiner |
| boy | NN | Noun |
| is | VBZ | Verb |
| playing | VBG | Verb (Present Participle) |
| cricket | NN | Noun |


Tokenization – Breaks sentences into words.

Stopword Removal – Removes selected common words.

Stemming – Removes word endings to obtain stems.

Lemmatization – Converts words into meaningful base forms

POS Tagging – Identifies the grammatical role of each word.


```python
paragraph = """Natural Language Processing (NLP) is a branch of Artificial Intelligence that helps computers understand and process human language. During text preprocessing, stopwords are common words such as is, am, are, the, a, an, in, on, and, of, and to, which often carry less meaningful information in certain NLP tasks. Stopword removal is the process of removing these words from text to reduce unnecessary data and focus on important words. For example, the sentence "I am learning the concepts of NLP" becomes "learning concepts NLP" after removing stopwords. This helps reduce the size of the dataset and makes text processing more efficient. However, stopwords should not always be removed because words like "not" and "don't" can completely change the meaning of a sentence."""
```

```python
import nltk

nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.corpus import stopwords
sentences = nltk.sent_tokenize(paragraph)
```

```python
sentences
```

```python

for i in range(len(sentences)):
    sentences[i] = sentences[i].lower()
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word not in set(stopwords.words('english'))]
    # sentences[i] = " ".join(words)
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
```


## 3. Difference Between `pos_tag()` and `pos_tag_sents()`

| Feature | `pos_tag()` | `pos_tag_sents()` |
|---|---|---|
| Purpose | Tags one sentence | Tags multiple sentences |
| Input | List of words | List of lists of words |
| Output | List of tagged words | List of tagged sentences |
| Use Case | Single sentence | Multiple sentences / documents |


```python
import nltk
from nltk.tokenize import word_tokenize

text = "Taj Mahal is a beautiful Monument. It is located in Agra."

# Step 1: Split text into sentences
sentences = nltk.sent_tokenize(text)

# Step 2: Tokenize each sentence
tokens = [word_tokenize(sentence) for sentence in sentences]

# Step 3: POS Tagging
pos_tags = nltk.pos_tag_sents(tokens)

print(pos_tags)
```
