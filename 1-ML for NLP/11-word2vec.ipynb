# Word2Vec in NLP

Word2Vec (Word to Vector) is a word embedding technique in Natural Language Processing (NLP) that converts words into dense numerical vectors (embeddings).

Unlike Bag of Words and TF-IDF, Word2Vec captures semantic meaning and relationships between words by learning from their surrounding context.

For example, Word2Vec can learn that words like `King` and `Queen` have similar contextual meanings.

## 1. Why do we need Word2Vec?

In traditional techniques:

- **Bag of Words:** Counts how many times words appear.
- **TF-IDF:** Measures the importance of words.
- Both methods do not understand the semantic relationships between words.

Word2Vec solves this problem by:

1. Representing words as dense numerical vectors.
2. Capturing semantic relationships between words.
3. Identifying similarities between words.
4. Learning word representations based on surrounding context.

## 2. How does Word2Vec work?

Word2Vec learns the meaning of words by examining the words that appear around them.

Consider this sentence:

`I love machine learning`

Word2Vec learns relationships between words based on their surrounding context.

For example:
- `machine` appears near `learning`.
- `love` appears near `machine`.
- `I` appears near `love`.

After training, each word receives a numerical vector.

**Example (illustrative vectors):**

| Word | Vector |
|---|---|
| King | [0.25, 0.81, 0.42] |
| Queen | [0.24, 0.79, 0.45] |
| Apple | [0.91, 0.12, 0.32] |

King and Queen have similar vectors because their learned representations capture related contexts.

## 3. Types of Word2Vec

Word2Vec has two main architectures:

### 1. CBOW (Continuous Bag of Words)

CBOW predicts the **target word using its surrounding context words**.

Example:

`I love ____ learning`

The model tries to predict the missing word: `machine`.

**Working:**

Context words → Target word

Example:

Input: `I`, `love`, `learning`

Output: `machine`

### 2. Skip-gram

Skip-gram does the opposite. It predicts the **surrounding context words using a target word**.

Example:

`I love machine learning`

Input word: `machine`

The model tries to predict surrounding words such as `love` and `learning`.

**Working:**

Target word → Context words

Example:

Input: `machine`

Output: `love`, `learning`

## 4. CBOW vs Skip-gram

| Feature | CBOW | Skip-gram |
|---|---|---|
| Input | Context words | Target word |
| Output | Target word | Context words |
| Prediction | Context → Target | Target → Context |
| Training | Generally faster | Generally slower |
| Best suited | Often frequent words | Often useful for rare words |

## 5. Advantages of Word2Vec

1. Captures semantic relationships between words.
2. Produces dense and relatively low-dimensional vectors.
3. Identifies similar words.
4. Provides more meaningful word representations than simple frequency-based methods.
5. Useful for many NLP applications.

## 6. Disadvantages of Word2Vec

1. A word generally gets one fixed embedding, regardless of its meaning in different contexts.
2. Requires training data.
3. Cannot naturally represent words that were not seen during training (OOV words).
4. Does not understand complete sentence meaning by itself.

## 7. Applications of Word2Vec

1. **Sentiment Analysis:** Understanding relationships between words in reviews.
2. **Search Engines:** Finding related words and improving search results.
3. **Text Classification:** Representing words as numerical vectors.
4. **Recommendation Systems:** Identifying related items based on textual descriptions.
5. **Machine Translation:** Learning relationships between words.

## 8. Bag of Words vs TF-IDF vs Word2Vec

| Feature | BoW | TF-IDF | Word2Vec |
|---|---|---|---|
| Approach | Word frequency | Word importance | Word embeddings |
| Semantic meaning | No | No | Yes, learned from context |
| Vector type | Sparse | Sparse | Dense |
| Word relationships | No | No | Yes |
| Context | No | No | Local context during training |

## 9. Important for Exams: CBOW and Skip-gram

Understand both architectures and their differences.

**CBOW (Continuous Bag of Words):**
- Takes surrounding context words as input.
- Predicts the missing target word.
- Example: `I love ____ learning` → `machine`.

**Skip-gram:**
- Takes the target word as input.
- Predicts the surrounding context words.
- Example: `machine` → `love`, `learning`.

**Most important difference:**

CBOW predicts the target word from context, whereas Skip-gram predicts context words from the target word.

**Remember:**

- CBOW = Context → Target
- Skip-gram = Target → Context

## Conclusion

Word2Vec is a word embedding technique that converts words into dense numerical vectors by learning from their surrounding context.

- **BoW** = Counts words.
- **TF-IDF** = Measures word importance.
- **Word2Vec** = Learns word representations and semantic relationships.

Word2Vec has two main architectures: CBOW and Skip-gram. Both learn word embeddings, but they differ in how they use context to make predictions.
