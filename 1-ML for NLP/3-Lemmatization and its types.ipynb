Lemmatization in NLP (Natural Language Processing) is the process of converting a word into its meaningful dictionary base form, called a lemma.

Ex: Playing -> Play

Stemming: Studies → Studi (Rule-based removal)

Lemmatization: Studies → Study (Vocabulary and grammar)

When should we use them?

- Stemming: When speed matters and exact linguistic correctness is less important.
- Lemmatization: When meaningful words and linguistic accuracy matter.

```python
words = ["playing", "played", "studies", "children", "mice"]
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()
for word in words:
    print(lemmatizer.lemmatize(word))
```

pos

Noun -> n  
verb -> v  
adjective -> a  
adverb -> r  

```python
lemmatizer.lemmatize("going",pos="v")
```
