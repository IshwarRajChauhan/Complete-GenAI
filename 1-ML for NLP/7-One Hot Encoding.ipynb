
# One-Hot Encoding in NLP

**One-Hot Encoding** is a technique used to convert categorical data (words or labels) into numerical vectors containing only `0` and `1`.

In simple words, it converts each unique word into a binary vector so that machine learning algorithms can process it.

## 1. How Does One-Hot Encoding Work?

Suppose we have three words:

```python
words = ["Apple", "Banana", "Mango"]
```

First, create a vocabulary of unique words:

```text
Apple, Banana, Mango
```

Now, assign each word a binary vector:

| Word | Apple | Banana | Mango |
|---|---|---|---|
| Apple | 1 | 0 | 0 |
| Banana | 0 | 1 | 0 |
| Mango | 0 | 0 | 1 |

Each word gets:
- `1` at its corresponding position.
- `0` at all other positions.

For example:
- Apple → `[1, 0, 0]`
- Banana → `[0, 1, 0]`
- Mango → `[0, 0, 1]`

## 2. Python Implementation Using Scikit-learn

### Understanding the Functions

- `OneHotEncoder()` → Creates the encoder.
- `fit()` → Learns the unique categories.
- `transform()` → Converts categories into binary vectors.
- `fit_transform()` → Performs both operations together.

## 3. Advantages of One-Hot Encoding

1. Converts categorical data into numerical form.
2. Easy to understand and implement.
3. Does not create artificial rankings between categories.
4. Useful for machine learning algorithms.

## 4. Disadvantages of One-Hot Encoding

1. Creates large vectors when vocabulary is large.
2. Requires more memory for large datasets.
3. Does not understand the meaning of words.
4. Cannot capture relationships between similar words.

For example:

- King → `[1, 0, 0]`
- Queen → `[0, 1, 0]`

One-hot encoding cannot understand that King and Queen are related words.

## 5. One-Hot Encoding vs Word Embeddings

| Feature | One-Hot Encoding | Word Embeddings |
|---|---|---|
| Representation | Binary vectors | Dense numerical vectors |
| Values | 0 and 1 | Decimal values |
| Meaning | Does not capture meaning | Can capture semantic relationships |
| Vector size | Vocabulary size | Fixed dimensions |
| Similarity | No semantic similarity | Similar words can have similar vectors |

## 6. Conclusion

One-Hot Encoding is a technique used in NLP and machine learning to convert categorical words or labels into numerical binary vectors. Each unique category is represented by a vector containing one `1` and the remaining values as `0`. It allows machine learning algorithms to process categorical information but does not capture semantic meaning or relationships between words.


````python
from sklearn.preprocessing import OneHotEncoder
````

````python
genres = [
    ["Action"],
    ["Comedy"],
    ["Horror"],
    ["Romance"],
    ["Action"]
]

encoder = OneHotEncoder(sparse_output=False)
````

````python
print(genres)
````

````python
# Apply One-Hot Encoding
encoded_data = encoder.fit_transform(genres)
encoded_data
````
