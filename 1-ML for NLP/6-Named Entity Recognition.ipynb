
# Named Entity Recognition (NER) in NLP

**Named Entity Recognition (NER)** is a technique in Natural Language Processing (NLP) that identifies and classifies important entities in a sentence into predefined categories such as names of people, organizations, locations, dates, money, and more.

In simple words, NER helps computers identify **who, where, when, and what** is mentioned in a sentence.

## 1. Types of Named Entities

| Entity | Meaning | Example |
|---|---|---|
| PERSON | Name of a person | Elon Musk, Rahul |
| ORGANIZATION (ORG) | Company or org
anization | Google, Microsoft |
| LOCATION (GPE) | Country, city, or state | India, Delhi, London |
| DATE | Date or day | 12 November, Monday |
| TIME | Time | 10 AM, 5:30 PM |
| MONEY | Monetary value | $500, ₹5000 |
| PERCENT | Percentage | 50%, 90% |
| FACILITY | Buildings or infrastructure | Taj Mahal, Airport |

## 2. Example of NER

Consider this sentence:

**"Elon Musk founded Tesla in California in 2003."**

After Named Entity Recognition:

| Entity | Category |
|---|---|
| Elon Musk | PERSON |
| Tesla | ORGANIZATION |
| California | GPE |
| 2003 | DATE |

NER identifies important information and categorizes it automatically.


## 4. Why Do We Use NER?

1. Identifies names of people, companies, and locations.
2. Extracts useful information from large amounts of text.
3. Helps in information extraction and text analysis.
4. Used in chatbots and virtual assistants.
5. Helps organize unstructured text into structured data.

## 5. Applications of NER

- Search Engines
- Chatbots
- Resume Information Extraction
- News Analysis
- Medical Information Extraction
- Question Answering Systems
- Document Processing

## 6. Difference Between POS Tagging and NER

| Feature | POS Tagging | NER |
|---|---|---|
| Purpose | Identifies grammatical roles | Identifies named entities |
| Output | Noun, Verb, Adjective, etc. | Person, Organization, Location, etc. |
| Example | Delhi → Noun | Delhi → Location |
| Focus | Grammar | Important real-world entities |

```python

sentence = """The Eiffel Tower was built from 1887 to 1889 by French engineer
Gustave Eiffel, whose company specialized in building metal frameworks and structures."""

"""
Person Eg: Krish C Naik
Place Or Location Eg: India
Date Eg: September, 24-09-1989
Time Eg: 4:30pm
Money Eg: 1 million dollar
Organization Eg: iNeuron Private Limited
Percent Eg: 20%, twenty percent
"""
```

```python
import nltk
nltk.download('maxent_ne_chunker')
nltk.download('words')
words = nltk.word_tokenize(sentence)
```

```python
tag_elements = nltk.pos_tag(words)
tag_elements
```

```python
tree = nltk.ne_chunk(tag_elements)
tree.pretty_print()
```

```python
tree.draw()
```
